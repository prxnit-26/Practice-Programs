```python
### bar graph, pie graph and line graph in python
import pandas as pd
df = pd.read_csv(r"C:\Users\admin\Downloads\Titanic-Dataset (3).csv")
## bar graph
import matplotlib.pyplot as plt
# count passengers in each class 
counts = df["Pclass"].value_counts().sort_index()
# creating the bar graph
plt.bar(counts.index, counts.values)
# add names for everything
plt.title("Number of passengers as per class")
plt.xlabel("class name")
plt.ylabel("number of passemgers")
plt.xticks([1,2,3])
plt.show()
```

```python
# creating a bar graph with 
counts = df.groupby(["Survived", "Sex"]).size().unstack()
print(counts)
counts.plot(kind = "bar")
plt.xlabel("Survival Status")
plt.ylabel("Number of Passengers")
plt.title("Survival by gender")
plt.xticks([0,1], ["Not survived", "Survived"], rotation = 0)
plt.show()
```

```python
# scatter graph
data = df.dropna(subset = ["Fare", "Age"])
plt.scatter(data["Age"], data["Fare"])
plt.xlabel("Age")
plt.ylabel("Fare")
plt.title("Age vs Fare")
plt.show()
```

```python
# scatter graph with grid
data = df.dropna(subset = ["Fare", "Age"])
plt.scatter(data["Age"], data["Fare"])
plt.xlabel("Age")
plt.ylabel("Fare")
plt.title("Age vs Fare")
plt.grid(True)
plt.show()
```

```python
# line graph
avg_fare = df.groupby("Pclass")["Fare"].mean()
plt.plot(avg_fare.index, avg_fare.values, marker = "o")
plt.xlabel("Passenger Class")
plt.ylabel("Avg Fare")
plt.title("Average fare by class")
plt.show()
```
