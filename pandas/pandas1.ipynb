```python
#Creating a series
import pandas as pd
a = pd.Series([2,3,4])
print(a)
print(type(a))
```

```python
# creating ur own indices
a = pd.Series([2,3,4], index = ["b","c","d"])
print(a)
```

```python
# naming ur series (displayed while printing)
a = pd.Series([2,3,4], index=["a","b","c"], name="Series 3")
print(a)
```

```python
# creating a series from a dictionary
a = pd.Series({"a":1,"b":2,"c":3})
print(a)
b = pd.Series({1:"a",2:"b",3:"c"})
print(b)
```

```python
# data frame (data frames allow us to create a more complex tanle than the previous method)
a = {
"student names": ["Pranit", "Philip", "Aryan"],
"marks": [96,97,98]
}
b = pd.DataFrame(a)
print(b)
```

```python
# locating rows
# the loc keyword is used which stands for location.
a = {
"student names": ["Pranit", "Philip", "Aryan"],
"marks": [96,97,98]
}
b = pd.DataFrame(a)
print(b.loc[0]) #locating 1st row
print(b.loc[[0,1]])
```

```python
# creating ur own primary indices
a = {
"student names": ["Pranit", "Philip", "Aryan"],
"marks": [96,97,98]
}
b = pd.DataFrame(a, index=["S1","S2","S3"])
print(b)
```

```python
## cleaning of data
# cleaning of data (cleaning involves removing redundant columns such as "birth year and age", filling in missing values for ex "in temporary address 
# fill in the columns left blank by local ppl", arranging all data into a consistent format such as "first,middle, then last name", giving meaningful
# names to columns such as "replacing x,y,z with student names, marks, PRN")
a = {
"A":[1,2,3,None],
"B":[1,None,None,4],
"C":[1,2,3,4]
}
b = pd.DataFrame(a)
print("original data:")
print(b) # note that "None" is displayed as "NaN"
c = b.dropna() # remove entire row with even a single none value
print("new data:")
print(c)
d = b.fillna(0, inplace=True) # fillna() used to replace none value with smt else
print("after using fillna:")
print(d)
# b.fillna(b.mean(), inplace=True) # mean function() can be used to replace none values in a column with the mean of that specific column
print("after using fillna and mean:")
print(b.fillna(b.mean(), inplace=True))
```

```python
#understand what mean() actually does
a = {
"A":[1,2,3,None],
"B":[1,None,None,4],
"C":[1,2,3,4]
}
b = pd.DataFrame(a)
print(b.mean())
```
