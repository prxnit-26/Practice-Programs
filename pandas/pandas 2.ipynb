```python
# using to_string function to remove indices
import pandas as pd
a = {
    "A":[1,2,2,4,3,4,3,3],
    "B":[5,6,6,6,7,8,3,3]
}
b = pd.DataFrame(a)
print(b.to_string(index=False))
```

```python
# displaying duplicate values
c = b[b.duplicated()]
print(c.to_string(index=False)) #note that the duplicated function only displays 'corresponding duplicate values'
```

```python
# removing duplicate values
d = b.drop_duplicates(subset=['A'], keep='first', inplace=True)
print(d)
```

```python
# renaming column names to more meaningful ones
e = b.rename(columns={"A":"Column 1", "B":"Column 2"}, inplace=True)
print(e.to_string(index=False))
```
