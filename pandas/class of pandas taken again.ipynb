```python
import pandas as pd
A = {
    'A':[1,2,2,3,3,4],
    'B':[5,6,6,7,8,8]
}
B = pd.DataFrame(A)
print(B.to_string(index=False))
```

```python
X = B[B.duplicated()]
print(X.to_string(index=False))
```

```python
B.drop_duplicates(subset=['A'], keep='first', inplace=True)
print(B.to_string(index=False))
```

```python
B.rename(columns={"A":"column 1", "B":"column 2"}, inplace=True)
print(B.to_string(index=False))
```
