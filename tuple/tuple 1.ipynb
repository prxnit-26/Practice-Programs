```python
# to print nth element from the start and end
T = (10,20,30,40,50,60,70)
n = int(input("enter: "))
print("nth element from the start:", T[n-1])
print("nth element from the end:", T[-n])
```

```python
# to check whether an element exists in a tuple
T = (10,20,30,40,50,60,70)
element = int(input("enter element: "))
if element in T:
    print("element exists")
elif element not in T:
    print("element does not exist")
```

```python
# to convert list to tuple
L = [1,2,3,4]
T = tuple(L)
print("list:", L)
print("tuple:", T)
```

```python
# to find index of an element in the tuple
T = (1,2,3,4)
e = int(input("enter element: "))
if e in T:
    print("index:", T.index(e))
elif e not in T:
    print("element not found")
```

```python
# to replace last value in all tuples of a master list to 100
L1 = [(10,20,40), (40,50,60), (70,80,90)]
L2 = []
for i in L1:
    new_T = i[:-1] + (100,)
    L2.append(new_T)
print("before:", L1)
print("after:", L2)
```

```python
# to find minimum and maximum element in a tuple
T = (10,20,30,40)
minn = min(T)
maxx = max(T)
print("min element:", minn)
print("max element:", maxx)
```

```python
# to swap the first and last element of a tuple 
T = (10,20,30,40,50)
new_T = (T[-1],)+(T[1:-1])+(T[0],)
print("before:", T)
print("after:", new_T) 
```
