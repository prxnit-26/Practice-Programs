```python
# to search whether a key is present in the dictionary 
D = {1:"Pranit", 2:"Advika", 3:"Shandra", 4:"Philip"}
x = int(input("enter key u want to search for: "))
if x in D.keys():
    print('present')
else:
    print('absent')
```

```python
# to get the sum of all values in the dictionary
D = {"Pranit":96, "Advika":99, "Philip":100}
x = D.values()
print(type(x))
L = []
for i in x:
    L.append(i)
print("sum of values:", sum(L))
```

```python
# to print keys then values then both keys and values 
D = {1:"Pranit", 2:"Advika", 3:"Shandra", 4:"Philip"}
for key in D: # js note that for i in D: print(i),,, the output will be 1,2,3,4
    print(key)
for value in D.values():
    print(value)
for key,value in D.items():
    print(key,value)
```

```python
# to generate a dictionary w numbers from 1 to 15 as keys and values as the square of those nos.
D = {}
for i in range(1,16):
    D[i] = i**2
print(D)
```

```python
# to add and delete student information and then display student information
my_dict = {"Pranit":94, "Manthan":95, "Advika":96, "Philip":97, "Rachit":98}
print(my_dict)
my_dict["Asmita"] = 99
print(my_dict)
del(my_dict["Pranit"])
print(my_dict)
for k,v in my_dict.items():
    print("key:",k,",","value:",v)
```

```python
# to convert list to dictionary
L1 = [1, 2]
L2 = ["Pranit", "Philip"]
D = dict(zip(L1,L2))
print(D)
```

```python
# to convert dictionary to list
D = {"Pranit":94, "Manthan":95, "Advika":96, "Philip":97, "Rachit":98}
L1 = list(D.keys())
L2 = list(D.values())
print("keys:", L1)
print("values:", L2)
```

```python
# to get mean of all values
D = {"Pranit":94, "Manthan":95, "Advika":96, "Philip":97, "Rachit":98}
x = sum(D.values())
n = len(D)
mean = x/n
print("mean of values:", mean)
```

```python
# to sort the dictionary keys
D = {"Pranit":94, "Manthan":95, "Advika":96, "Philip":97, "Rachit":98}
x = sorted(D)
print("original dict:", D)
print("sorted keys list is:", x)
```
