```python
# To access elements of a list and append an element to a list
l = [1, 2, 3, 4, 5]

print("first elements is:", l[0])
print("third element is:", l[2])

l.append(6)

print("after appending, list is:", l)
# To reverse a list and count the occurrences of an element in a list
L1 = [1, 2, 3, 6, 4, 5, 6, 6]

L1.reverse()

print("reversed list is:", L1)

countt = L1.count(6)

print("count of 6 is:", countt)
# To append the elements of list 2 to list 1 in the front
L1 = [10, 20, 30]
L2 = [40, 50, 60]

L = L2 + L1

print("new list is:", L)
# Inserting an element
L = ['a', 'c', 'd']

L.insert(1, 'b')

print("List after inserting element:", L)
# Popping an element
L = ['Pranit', 'Advika', 'Philip']

L.pop(1)

print("updated list is:", L)
# Removing an element
L = ['Pranit', 'Advika', 'Philip']

L.remove('Pranit')

print("updated list is:", L)

# Accept 20 values from the user, then:
# (a) Display indices of all elements whose count is greater than 1
# (b) Display even and odd numbers
# (c) Display positive and negative numbers

L = []
for i in range(20):
    num = int(input("enter number: "))
    L.append(num)

# (a)
for i in range(20):
    if L.count(L[i]) > 1:
        print(i)

# (b)
print()

count_e = 0
count_o = 0

for i in L:
    if i % 2 == 0:
        count_e += 1
    elif i % 2 != 0:
        count_o += 1

print("count of even nos. is:", count_e)
print("count of odd nos. is:", count_o)

# (c)
count_p = 0
count_n = 0

for i in L:
    if i > 0:
        count_p += 1
    elif i < 0:
        count_n += 1

print("count of positive nos. is:", count_p)
print("count of negative nos. is:", count_n)


# Accept 10 values from the user and save them to a list.
# Perform the following operations on the list:
# (a) Display list in ascending order
# (b) Display list in descending order
# (c) Display length of list

L = []

for i in range(10):
    num = int(input("enter number: "))
    L.append(num)

# (a)
L.sort()

print("list in asc order is:", L)

# (b)
L.sort(reverse=True)

print("list in desc order is:", L)

# (c)
n = len(L)

print("length of list is:", n)


# Accept 2 lists from the user and merge them into a single list

print("for List1: ")

L1 = []

for i in range(3):
    num = int(input("enter element: "))
    L1.append(num)

print("for List2: ")

L2 = []

for i in range(3):
    num = int(input("enter element: "))
    L2.append(num)

print(L1, L2)

L = L1 + L2

print("merged list is:", L)
```
